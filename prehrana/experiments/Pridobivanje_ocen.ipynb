```python
import requests

api_key = 'YOUR_API_KEY'
base_url = 'https://maps.googleapis.com/maps/api/place/textsearch/json'
query = 'Gregorčičeva ulica 3, 1000 Ljubljana, Slovenia'  # Zamenjaj z imenom in naslovom
url = f'{base_url}?query={query}&key={api_key}'

response = requests.get(url)
data = response.json()

if data['results']:
    place_id = data['results'][0]['place_id']
    print(f"Place ID: {place_id}")
```

```python
details_url = 'https://maps.googleapis.com/maps/api/place/details/json'
details_query = f'{details_url}?place_id={place_id}&key={api_key}'
details_response = requests.get(details_query)
details_data = details_response.json()

if 'result' in details_data:
    ratings = details_data['result'].get('rating', 'No rating available')
    print(f"Rating: {ratings}")
```

```python
from thefuzz import process

# Seznam restavracij
restaurants = ["Znanstvena kavarna Mafija", "Restavracija Menza IJS", "Nova Restavracija"]

# Uporabniški vnos
user_input = "Menza IJS"

# Poišči najbolj podobno ime
best_match, score = process.extractOne(user_input, restaurants)

# Nastavi prag za sprejemljivo ujemanje (npr. 80%)
if score > 80:
    print(f"Najdeno ujemanje: {best_match} (zanesljivost: {score}%)")
else:
    print("Ni dovolj dobrega ujemanja.")
```
